```python
X = df.iloc[:,0: 29]
y= df["target"]
```

#Varience Threshold method -
**Removing features with low variance** - This feature selection algorithm looks only at the features (X), not the desired outputs (y), and can thus be used for unsupervised learning.

It removes all features whose variance doesn’t meet some threshold. By default, it removes all zero-variance features, i.e. features that have the same value in all samples.

```python

from sklearn.feature_selection import VarianceThreshold
```

```python
data = VarianceThreshold(threshold=4)
data.fit_transform(X)
```

above output gives the columns with variance grater than the threshold.

```python
print(f"Variances of individual features:\n {data.variances_}")
```

```python
print(f"Number of features seen during fit: {data.n_features_in_}\n")
print(f"Names of features seen during fit: \n{data.feature_names_in_}")
```

```python
print("feature names according to selected features")
print(data.get_feature_names_out())
```

```python
print("Iinteger index of the features selected")
print(data.get_support(indices=True))
```

#Correlation Coefficient

 Measures linear relationship between two continuous variables.

```python
correlation_scores = X.corrwith(y)
#print(correlation_scores)
top_features = correlation_scores.abs().sort_values(ascending=False).head(10)
print(f"top 10 features:\n{top_features}")
```

```python

print("\nCorrelation scores:")
print(correlation_scores.loc[top_features.index])
```

#Univariate feature selection

Univariate feature selection works by selecting the best features based on univariate statistical test.

SelectKBest(score_func=<function f_classif>, k=10) - Feature selection algorithm to select features according to the k highest scores.

```python
from sklearn.feature_selection import SelectKBest, f_classif

anova_selector = SelectKBest(score_func=f_classif,k=10)

X_anova = anova_selector.fit_transform(X, y)
```

```python
print(f"Selected features (ANOVA):\n {X_anova}")
```

```python
print(f" Scores of features: \n{anova_selector.scores_}")
```

```python
print(f"selected feature names: {anova_selector.get_feature_names_out()}")
```

#Recursive feature elimination (RFE)

RFE is to select features by recursively considering smaller sets of features.

First, the estimator is trained on the initial set of features and the importance of each feature is obtained either through any specific attribute.

Then, the least important features are pruned from the current set of features. That procedure is recursively repeated on the pruned set until the desired number of features(n_features_to_select) to select is eventually reached.

**n_features_to_selectint**- The number of features to select.

If None-  half of the features are selected.

If integer- the parameter is the absolute number of features to select.

If float between 0 and 1- it is the fraction of features to select.

**step** - If greater than or equal to 1 - then the integer number of features to remove at each iteration.

If within (0.0, 1.0)-  then step corresponds to the percentage rounded down of features to remove at each iteration.

**importance_getter** - feature importance of which features would remove from the set.



```python
from sklearn.feature_selection import RFE
from sklearn.svm import SVC

estimator = SVC(kernel="linear")
selector = RFE(estimator, n_features_to_select=5, step=1)
selector = selector.fit(X, y)
```

```python
print(f"The number of selected features: {selector.n_features_}")
print(f"The selected features: {selector.support_}")
print(f"The ranking of the features: {selector.ranking_}")
```

```python
print(f"Names of features seen during fit: {selector.feature_names_in_}")
```

```python

print(f"feature names according to selected features:\n {selector.get_feature_names_out(input_features=selector.feature_names_in_)}")

```
